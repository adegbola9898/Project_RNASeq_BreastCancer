```python
import pandas as pd

# Load the CSV file
file_path = '/content/output.csv'  # Replace with your actual file path
data = pd.read_csv(file_path, header=None)  # No header in the file

# Extract the first row as a list
gsm_files = data.iloc[0].dropna().tolist()  # Drop any NaN values, if present

# Flatten the list (if it's a single row split into columns)
gsm_files_flat = [item for sublist in gsm_files for item in str(sublist).split(',')]

# Remove empty strings (in case there's a trailing comma)
gsm_files_clean = [gsm.strip() for gsm in gsm_files_flat if gsm.strip()]

# Convert to a pandas Series to check for duplicates
gsm_series = pd.Series(gsm_files_clean)

# Check for duplicate entries
duplicates = gsm_series[gsm_series.duplicated()]
unique_count = gsm_series.nunique()

if duplicates.empty:
    print(f"All GSM files are unique. There are {unique_count} unique GSM files.")
else:
    print(f"There are duplicates in the list. Total unique GSM files: {unique_count}")
    print(f"Duplicate entries:\n{duplicates.value_counts()}")
```

```python
import pandas as pd

# Load the first dataset (with unique accession numbers)
file_path_gsm = '/content/output.csv'  # Path to your GSM file
gsm_data = pd.read_csv(file_path_gsm, header=None)

# Extract unique GSM accession numbers
gsm_files = gsm_data.iloc[0].dropna().tolist()
gsm_files_flat = [item for sublist in gsm_files for item in str(sublist).split(',')]
gsm_files_clean = [gsm.strip() for gsm in gsm_files_flat if gsm.strip()]

# Convert to a set for faster filtering
gsm_set = set(gsm_files_clean)

# Load the second dataset
file_path_second = '/content/Untitled spreadsheet - Sheet1.csv'
second_data = pd.read_csv(file_path_second)

# Filter rows where 'Accession' is in the GSM set
filtered_data = second_data[second_data['Accession'].isin(gsm_set)]

# Save the filtered dataset (optional)
filtered_output_path = '/content/filtered_data.csv'
filtered_data.to_csv(filtered_output_path, index=False)

# Display the filtered data
print(filtered_data.head())
```

```python
# Summarize the filtered data by the Treatment column
treatment_summary = filtered_data['Treatment'].value_counts()

# Convert to a DataFrame for better readability
treatment_summary_df = treatment_summary.reset_index()
treatment_summary_df.columns = ['Treatment', 'Count']

# Display the summary
print(treatment_summary_df)
```

```python
# Filter the data for only "Pre-treatment" samples
pre_treatment_data = filtered_data[filtered_data['Treatment'] == 'Pre-treatment']

# Save the filtered dataset (optional)
pre_treatment_output_path = '/content/pre_treatment_data.csv'
pre_treatment_data.to_csv(pre_treatment_output_path, index=False)

# Display the first few rows of the filtered data
print(pre_treatment_data.head())
```

```python
patient_summary = pre_treatment_data['Patient id'].value_counts()
print(patient_summary)
```

```python
# Group by Patient id and count the number of rows (samples) per patient
patient_sample_counts = pre_treatment_data.groupby('Patient id').size()

# Count the number of patients with two samples and one sample
patients_with_two_samples = patient_sample_counts[patient_sample_counts == 2].count()
patients_with_one_sample = patient_sample_counts[patient_sample_counts == 1].count()

# Print the results
print(f"Number of patients with two pretreatment samples: {patients_with_two_samples}")
print(f"Number of patients with only one pretreatment sample: {patients_with_one_sample}")
```

```python
import pandas as pd

# Load the preprocessed datasets
file_path1 = '/content/GSE52194_preprocessed1.tsv'
file_path2 = '/content/GSE58135_preprocessed.tsv'
file_path3 = '/content/GSE243375_preprocessed(1).tsv'

# Load the datasets
data1 = pd.read_csv(file_path1, sep='\t', index_col=0)
data2 = pd.read_csv(file_path2, sep='\t', index_col=0)
data3 = pd.read_csv(file_path3, sep='\t', index_col=0)

# Convert indices to strings to avoid float conversion
data1.index = data1.index.astype(str)
data2.index = data2.index.astype(str)
data3.index = data3.index.astype(str)

# Compare row names to check if they are identical across datasets
rows_data1 = data1.index
rows_data2 = data2.index
rows_data3 = data3.index

if rows_data1.equals(rows_data2) and rows_data2.equals(rows_data3):
    print("All three datasets have identical row names.")
else:
    print("The datasets do not have identical row names.")

    # Find and print differences between the datasets if they exist
    print("\nDifferences between GSE52194 and GSE58135 row names:")
    diff_1_2 = rows_data1.symmetric_difference(rows_data2)
    print(diff_1_2[:20].tolist() if len(diff_1_2) > 0 else "No differences")

    print("\nDifferences between GSE58135 and GSE243375 row names:")
    diff_2_3 = rows_data2.symmetric_difference(rows_data3)
    print(diff_2_3[:20].tolist() if len(diff_2_3) > 0 else "No differences")

    print("\nDifferences between GSE52194 and GSE243375 row names:")
    diff_1_3 = rows_data1.symmetric_difference(rows_data3)
    print(diff_1_3[:20].tolist() if len(diff_1_3) > 0 else "No differences")
```

```python
import pandas as pd

# File paths
file_path1 = '/content/GSE52194_preprocessed1.tsv'
file_path2 = '/content/GSE58135_preprocessed.tsv'
file_path3 = '/content/GSE243375_preprocessed_cleaned.tsv'

# Load the datasets
data1 = pd.read_csv(file_path1, sep='\t', index_col=0)
data2 = pd.read_csv(file_path2, sep='\t', index_col=0)
data3 = pd.read_csv(file_path3, sep='\t', index_col=0)

# Standardize the indices of all datasets
def clean_index(df):
    # Convert indices to strings to avoid automatic conversion issues
    df.index = df.index.astype(str)
    # Remove '.0' suffix from row names, if present
    df.index = df.index.str.replace(r'\.0$', '', regex=True).str.strip()
    return df

# Apply cleaning function to all datasets
data1_cleaned = clean_index(data1)
data2_cleaned = clean_index(data2)
data3_cleaned = clean_index(data3)

# Save the cleaned versions back to files (optional)
data1_cleaned.to_csv('/content/GSE52194_preprocessed_cleaned.tsv', sep='\t')
data2_cleaned.to_csv('/content/GSE58135_preprocessed_cleaned.tsv', sep='\t')
data3_cleaned.to_csv('/content/GSE243375_preprocessed_cleaned_v2.tsv', sep='\t')

# Check for identical row names across the cleaned datasets
rows_data1 = data1_cleaned.index
rows_data2 = data2_cleaned.index
rows_data3 = data3_cleaned.index

if rows_data1.equals(rows_data2) and rows_data2.equals(rows_data3):
    print("All three datasets have identical row names after cleaning.")
else:
    print("The datasets still do not have identical row names.")

    # Print any remaining differences
    print("\nDifferences between GSE52194 and GSE58135 row names:")
    diff_1_2 = rows_data1.symmetric_difference(rows_data2)
    print(diff_1_2[:20].tolist() if len(diff_1_2) > 0 else "No differences")

    print("\nDifferences between GSE58135 and cleaned GSE243375 row names:")
    diff_2_3_cleaned = rows_data2.symmetric_difference(rows_data3)
    print(diff_2_3_cleaned[:20].tolist() if len(diff_2_3_cleaned) > 0 else "No differences")

    print("\nDifferences between GSE52194 and cleaned GSE243375 row names:")
    diff_1_3_cleaned = rows_data1.symmetric_difference(rows_data3)
    print(diff_1_3_cleaned[:20].tolist() if len(diff_1_3_cleaned) > 0 else "No differences")
```
